```python
import yfinance as yf
tw2330 = yf.download('2330.TW',start='2024-01-01',end='2024-06-01', auto_adjust=True)
tw2303 = yf.download('2303.TW',start='2024-01-01',end='2024-06-01', auto_adjust=True)
tw2454 = yf.download('2454.TW',start='2024-01-01',end='2024-06-01', auto_adjust=True)
tw2317 = yf.download('2317.TW',start='2024-01-01',end='2024-06-01', auto_adjust=True)
```

```python
tw2330.info()
```

```python
tw2330['Close']['2330.TW']
tw2330_data = tw2330[('Close','2330.TW')]
tw2303_data = tw2303[('Close','2303.TW')]
tw2454_data = tw2454[('Close','2454.TW')]
tw2317_data = tw2317[('Close','2317.TW')]
```

```python
import pandas as pd
data = {
    "2330.TW":tw2330_data,
    "2303.TW":tw2303_data,
    "2454.TW":tw2454_data,
    "2317.TW":tw2317_data
}
df = pd.DataFrame(data)
df
```

```python
import yfinance as yf

# 股票代碼列表
tickers = ['2330.TW', '2303.TW', '2454.TW', '2317.TW']

# 一次性下載所有股票資料
all_data = yf.download(tickers, start='2024-01-01', end='2024-06-01', auto_adjust=True)

# yfinance 會回傳一個包含 'Open', 'High', 'Low', 'Close' 等的多層次欄位 DataFrame
# 我們可以直接選取 'Close' 欄位來取得我們需要的 DataFrame
df_close = all_data['Close']

# 顯示整理好的 DataFrame
print(df_close.head())

# 直接繪圖
df_close.plot(title='Stock Closing Prices')
```
